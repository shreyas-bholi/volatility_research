```python
import numpy as np 
import pandas as pd
```

```python
df = pd.read_csv('data.csv')

df
```

```python
# 1. Ensure the column is datetime type
df['Date'] = pd.to_datetime(df['Date'])
# 2. Get the day of the week
# Option A: Day name (e.g., 'Monday', 'Tuesday')
df['day_name'] = df['Date'].dt.day_name()
# Option B: Day of week as an integer (0 = Monday, 6 = Sunday)
df['day_number'] = df['Date'].dt.dayofweek

df
```
